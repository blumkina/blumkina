## Задание 1. Логистическая регрессия своим спуском

Это тот же спуск, что на семинаре (часть 4 лабораторной), только на настоящих данных. Обозначения: $X$ — матрица признаков со столбцом единиц, $h=\sigma(X\theta)$ — вероятности класса 1, $\sigma(z)=\dfrac{1}{1+e^{-z}}$.

**Потеря** — средняя логистическая (её пишем в `losses`):

$$
Q(\theta) = -\frac1n\sum_{i=1}^{n}\Bigl[y_i\ln h_i + (1-y_i)\ln(1-h_i)\Bigr].
$$

Под логарифм добавьте `1e-12`, чтобы не получить $\ln 0$.

**Шаг спуска** — формула лекции. Градиент — **сумма** по объектам, без деления на $n$ и без множителя 2 (двойка была в MSE линейной регрессии из ДЗ 3, здесь её нет):

$$
\theta \leftarrow \theta - \eta\, X^{\mathsf T}\bigl(\sigma(X\theta) - y\bigr).
$$

**Старт** — $\theta = 0$ (вектор нулей). `losses` — потеря на старте и после каждого шага, всего `N_ITER + 1` чисел.

**Шаг и число шагов** заданы, их не меняйте: $\eta = 4\,/\,\lambda_{\max}(X^{\mathsf T}X)$ (наибольшее собственное число, `np.linalg.eigvalsh(...).max()`), `N_ITER = 5000`. Шаг вдвое больше, чем на занятии ($2/\lambda_{\max}$): так 5000 шагов хватает, чтобы дойти до минимума, а спуск остаётся устойчивым.

**Сравнение со scikit-learn.** Обучите `LogisticRegression(penalty=None, fit_intercept=False, max_iter=5000)` на тех же `Xb_train` (столбец единиц уже есть, поэтому `fit_intercept=False`). Напечатайте наибольшую разницу вероятностей на тесте $\max_i |p_i - p_i^{\text{sklearn}}|$ и ROC-AUC обеих моделей. Веса могут разойтись в третьем знаке, поэтому сравнивают вероятности.

```python
def sigmoid(z):
    # TODO: 1 / (1 + exp(-z)); аргумент сначала обрезать: z = np.clip(z, -30, 30)
    z = np.clip(z, -30, 30)
    return 1 / (1 + np.exp(-z))


def log_loss(X, y, theta):
    # TODO: средняя логистическая потеря (формула выше), под логарифм + 1e-12
    h = sigmoid(X @ theta)
    loss = y * np.log(h + 1e-12)
    loss = loss + (1 - y) * np.log(1 - h + 1e-12)
    return -np.mean(loss)


def logreg_gd(X, y, eta, n_iter):
    # TODO: theta = np.zeros(X.shape[1]); losses = [log_loss(X, y, theta)]
    # в цикле n_iter раз: h = sigmoid(X @ theta); grad = X.T @ (h - y); theta = theta - eta * grad;
    # после шага losses.append(log_loss(X, y, theta)).
    # Вернуть theta, np.array(losses)
    theta = np.zeros(X.shape[1])
    losses = [log_loss(X, y, theta)]   # потеря на старте
    for _ in range(n_iter):
        h = sigmoid(X @ theta)       # вероятности: сигмоида от X @ theta
        grad = X.T @ (h-y)    # градиент: X.T @ (h - y), без множителя 2
        theta = theta - eta * grad   # шаг: theta минус eta на градиент
        losses.append(log_loss(X, y, theta))
    return theta, np.array(losses)
```

```python
# Проверка формул (не меняйте эту ячейку): один шаг из лекции
_X = np.array([[1.0, 1.0],
               [1.0, 2.0],
               [1.0, 3.0]])
_y = np.array([0.0, 1.0, 1.0])
_theta, _losses = logreg_gd(_X, _y, eta=0.5, n_iter=1)
assert np.allclose(_theta, [0.25, 1.0], atol=1e-8), (
    f"один шаг лекции дал {_theta}, ожидается (0.25, 1). "
    "Градиент — сумма X.T @ (h - y), без множителя 2 и без деления на n.")
assert len(_losses) == 2, "losses должен содержать стартовое значение и значение после шага"
assert _losses[-1] < _losses[0], "после шага из лекции потеря должна уменьшиться"
print("Шаг из лекции совпал")
```

## Задание 3. Порог под полноту 0.95

Нужен самый высокий порог, при котором полнота на тесте не ниже 0.95. Более низкий порог тоже наберёт такую полноту, но добавит ложных тревог, поэтому берётся верхний. Относительно 0.5 он может и подняться, и опуститься: это зависит от разбиения. Функция ниже задаёт правило, её не меняйте.

**Важно: порог подбирается по тесту.** По этим же объектам дальше считаются precision и число ложных тревог, поэтому они получаются оптимистичными: порог подогнан под этот тест. В настоящей работе порог выбирают на отдельной валидационной части или кросс-валидацией, а тест трогают один раз в конце (это тема лабораторной 5). Здесь тест играет роль валидационной части, чтобы ваши числа можно было сверить с эталоном.

```python
def choose_threshold(y_true, prob, min_recall=0.95):
    """Самый высокий порог, при котором полнота не ниже min_recall.

    Кандидаты перебираются от 1 к 0, поэтому первый подходящий — максимальный.
    """
    candidates = np.concatenate([[1.0], np.sort(np.unique(prob))[::-1], [0.0]])
    for t in candidates:
        pred = (prob >= t).astype(int)
        if recall_score(y_true, pred, zero_division=0) >= min_recall:
            return float(t)
    return 0.0
```
